```python
matrix_A = [
    [1,2],[3,4]
]
matrix_B = [
    [5,6],[7,8]
]
result = [
    [0,0],[0,0]
]
for row in range(2):
  for col in range(2):
    result[row][col] = matrix_A[row][col] + matrix_B[row][col]
print("result of matrix Addition:")
for row in result:
  print(row)
```

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None


def inorder(root):
  if root is None:
    return
  inorder(root.left)
  print(root.data, end=" ")
  inorder(root.right)

root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
print("Inorder Traversal:")
inorder(root)
```
