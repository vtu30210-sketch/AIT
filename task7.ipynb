```python
# Monkey and Banana Problem

monkey_position = "door"
box_position = "corner"
banana_position = "center"
monkey_on_box = False
banana_obtained = False

print("Initial State:")
print("Monkey is at", monkey_position)
print("Box is at", box_position)
print("Banana is at", banana_position)

# Step 1: Monkey moves to the box
monkey_position = box_position
print("\n1. Monkey moves to the box.")

# Step 2: Monkey pushes the box below the banana
box_position = banana_position
monkey_position = banana_position
print("2. Monkey pushes the box below the banana.")

# Step 3: Monkey climbs onto the box
monkey_on_box = True
print("3. Monkey climbs onto the box.")

# Step 4: Monkey grabs the banana
if monkey_on_box and monkey_position == banana_position:
    banana_obtained = True
    print("4. Monkey grabs the banana.")

# Final state
if banana_obtained:
    print("\nGoal achieved: Monkey has the banana!")
else:
    print("\nGoal not achieved.")
```
